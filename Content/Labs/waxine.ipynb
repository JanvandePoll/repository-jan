# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

"""
Onderzoek naar het waxinekaarsje-onder-glas experiment

Situatie:
- Een waxinekaarsje brandt onder een omgekeerd glas.
- In variant 1 staat het kaarsje gewoon op tafel.
- In variant 2 drijft het kaarsje op water en wordt het glas over kaars + water gezet.
- In variant 2 stijgt het water in het glas nadat de kaars dooft.

Er worden drie verklaringen genoemd:
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af,
   waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de
   binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water
   omhoog gezogen.

Doel:
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.

Analyse per verklaring:

1. Zuurstofverbruik → vacuüm
   - Lucht bestaat voor ongeveer 21% uit zuurstof.
   - Tijdens het branden van de kaars wordt een deel van die zuurstof omgezet in CO₂.
   - De hoeveelheid gasdeeltjes verdwijnt niet volledig: O₂ wordt vervangen door CO₂.
   - Er is dus geen “echt vacuüm” en het totale gasvolume verandert maar heel weinig.
   - Zelfs als een aanzienlijk deel van de zuurstof wordt verbruikt, is de relatieve afname van het totale aantal
     gasdeeltjes klein (en dus ook de drukdaling).
   - Dit effect is te klein om de duidelijk zichtbare stijging van het water (meestal enkele centimeters) te verklaren.
    Conclusie: zuurstofverbruik alleen is geen goede hoofdverklaring.

2. Verwarming en afkoeling van de lucht (thermisch effect)
   - De kaars verwarmt de lucht in het glas.
   - Warme lucht zet uit. Omdat het glas niet perfect afsluit, kan een deel van de warme lucht ontsnappen.
   - Zolang de kaars brandt, is de lucht in het glas warmer dan de omgeving en blijft de druk ongeveer gelijk aan
     de buitenlucht (door uitstromen).
   - Wanneer de kaars dooft, koelt de lucht in het glas af naar omgevingstemperatuur.
   - Bij afkoeling krimpt het gasvolume (ideale gaswet: pV = nRT). Omdat het volume van het glas vast is en er
     geen lucht meer bij kan, daalt de druk in het glas.
   - Het water stijgt in het glas totdat het drukverschil tussen binnen en buiten in evenwicht is met de
     hydrostatische druk van de waterkolom.
   - Dit effect kan een drukdaling van de orde van 5–15% geven, wat overeenkomt met een waterstijging van
     ongeveer 1–3 cm, zoals vaak wordt waargenomen.
    Conclusie: dit is de belangrijkste en meest waarschijnlijke verklaring.

3. Condensatie van waterdamp
   - Bij verbranding van kaarsvet (een koolwaterstof) ontstaan CO₂ en H₂O (waterdamp).
   - De waterdamp condenseert tegen het koude glas en wordt vloeibaar.
   - Gecondenseerde waterdamp telt niet meer mee als gas, dus het aantal gasdeeltjes in het glas neemt af.
   - Minder gasdeeltjes betekent lagere druk (bij gelijk volume en temperatuur).
   - Dit effect werkt in dezelfde richting als de afkoeling: het verlaagt de druk in het glas.
   - De hoeveelheid waterdamp die condenseert is echter relatief klein, dus de bijdrage aan de totale drukdaling
     is meestal kleiner dan die van de temperatuurverandering.
    Conclusie: condensatie speelt een rol, maar is een secundair effect.

Eindconclusie:
- De meest waarschijnlijke hoofdverklaring is verklaring 2:
  De lucht in het glas wordt door de kaars verwarmd, zet uit en kan deels ontsnappen. Na het doven koelt de lucht af,
  krimpt het gas en daalt de druk in het glas. Hierdoor stijgt het water in het glas totdat het drukverschil is
  gecompenseerd.
- Verklaring 3 (condensatie van waterdamp) versterkt dit effect enigszins, maar is op zichzelf te klein om het
  volledige verschijnsel te verklaren.
- Verklaring 1 (zuurstofverbruik → vacuüm) is fysisch gezien geen goede hoofdverklaring, omdat de zuurstof wordt
  omgezet in andere gassen (zoals CO₂) en de netto afname van het aantal gasdeeltjes te klein is om de waargenomen
  waterstijging te veroorzaken.
"""
